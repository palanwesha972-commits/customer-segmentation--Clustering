Feature Engineering

```python
# Convert Dt_Customer to Date Format
data['Dt_Customer'] = pd.to_datetime(data['Dt_Customer'],format='%d-%m-%Y')
```

```python
data['Dt_Customer'].head()
```

```python
# Create to age feature
data['Age'] = 2015-data['Year_Birth']
```

```python
data['Age'].head()
```

```python
# Total-Spending
data['Total_Spending'] = (
    data['MntWines'] +
    data['MntFruits'] +
    data['MntMeatProducts'] +
    data['MntFishProducts'] +
    data['MntSweetProducts'] +
    data['MntGoldProds']
)
```

```python
data['Total_Spending'].head()
```

```python
# create Children
data['Children'] = data['Kidhome'] + data['Teenhome']
```

```python
data['Children'].head()
```

```python
# create Is_parent
data['Is_Parent'] = np.where(data['Children'] > 0, 1, 0)
```

Data Visualizations

```python
# Response Distribution
ax = sns.countplot(x='Response', data=data)
for container in ax.containers:
    ax.bar_label(container)
plt.title('Customer Response')
plt.show()
```

```python
# income distribution
ax = sns.histplot(data['Income'], bins=30)
plt.title('Income Distribution')
for p in ax.patches:
    ax.annotate(f'{p.get_height():.0f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom',fontsize=8)
plt.show()
```

```python
plt.figure(figsize=(10,5))
ax = sns.histplot(data[data['Total_Spending'] < 2000]['Total_Spending'], bins=30)
for p in ax.patches:
    if p.get_height() > 0:
        ax.text(p.get_x() + p.get_width()/2, p.get_height() + 5,
                int(p.get_height()),
                ha='center', va='bottom', fontsize=8)
plt.title('Total Spending Distribution')
plt.xlabel('Total_Spending')
plt.ylabel('Count')
plt.show()
```

```python
 plt.figure(figsize=(10,5))
ax = sns.histplot(data[data['Total_Spending'] < 2000]['Total_Spending'], bins=30)
for container in ax.containers:
    ax.bar_label(container, fontsize=8)
plt.title('Total Spending Distribution')
plt.show()
```

Clustering

```python
data.columns
```

```python
data['Children'] = data['Kidhome'] + data['Teenhome']
features = ['Income', 'Age', 'Total_Spending', 'Children']
X = data[features]
print(X.head())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
data['Cluster'] = kmeans.fit_predict(X_scaled)

print(data['Cluster'].value_counts())
```

```python
data[['Income', 'Age', 'Total_Spending', 'Children', 'Cluster']].head(10)
```

```python
sns.scatterplot(data=data, x='Income', y='Total_Spending', hue='Cluster')
plt.title('Customer Segmentation')
plt.show()
```

```python
cluster_profile = data.groupby('Cluster')[['Income', 'Age', 'Total_Spending', 'Children']].mean().round(1)
cluster_profile
```

```python
sns.scatterplot(data=data, x='Age', y='Total_Spending', hue='Cluster', palette='viridis', s=80)
plt.title('Customer Segmentation by Age & Spending')
plt.show()
```

```python
plt.figure(figsize=(13,8))
pl = sns.countplot(x=data['Cluster'])
pl.set_title('Distribution Of The Clusters')
plt.show()
```
